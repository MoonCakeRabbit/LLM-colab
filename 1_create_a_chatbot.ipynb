```python
from transformers import pipeline
```

```python
# !pip list | grep transformers
```

```python
chatbot = pipeline(task="text-generation",
                   model="facebook/blenderbot-400M-distill")
```

```python
user_message="""
What are some fun activities I can do in the winter?
"""
```

```python
# 将用户消息传递给chatbot并获取回复
response = chatbot(user_message, max_length=100, truncation=True)
print(response)
```

```python
bot_response = response[0]['generated_text']
print(bot_response)
```

```python
print(chatbot("What else do you recommend?")[0]['generated_text'])
```

```python
# 创建一个对话历史变量
conversation_history = ""

def chat_with_memory(user_input):
    global conversation_history

    # 如果有历史，将其与当前输入组合
    if conversation_history:
        full_input = f"{conversation_history}\nUser: {user_input}\nBot:"
    else:
        full_input = f"User: {user_input}\nBot:"

    # 调用模型获取响应
    response = chatbot(full_input, max_length=200, truncation=True)
    bot_response = response[0]['generated_text']

    # 从响应中提取模型回答部分（移除输入部分）
    model_answer = bot_response[len(full_input):].strip()

    # 更新对话历史
    conversation_history = f"{full_input} {model_answer}"

    return model_answer

# 第一次对话
user_message = "What are some fun activities I can do in the winter?"
print(f"User: {user_message}")
answer = chat_with_memory(user_message)
print(f"Bot: {answer}")

# 第二次对话（模型会记住之前的对话）
follow_up = "Which of those is good for kids?"
print(f"User: {follow_up}")
follow_up_answer = chat_with_memory(follow_up)
print(f"Bot: {follow_up_answer}")
```
