```python
from transformers import pipeline
import torch
```

```python
translator = pipeline(task="translation",
                      model="facebook/nllb-200-distilled-600M",
                      torch_dtype=torch.bfloat16)
```

```python
text="""\
My dog is cute,\
Your cat is beautiful.
Her panda is friendly.
His llama is thoughtful.\
We all have nice pets!"""
```

```python
text_translated=translator(text,src_lang="eng_Latn", tgt_lang="zho_Hans")
```

```python
summarizer=pipeline(task="summarization",
                    model="facebook/bart-large-cnn",
                    torch_dtype=torch.bfloat16)
```

```python
text = """How fortunate I’m to have such a beautiful view from my window!!
Looking at everyday to the sight of this sky and sea brings a sense of peace and serenity to my daily routine.
I want to show you the beauty of the nature where the sky meets the sea on the horizon that creates a breathtaking sight and make you amazed.
The combination of blues and light purples of the sky with the deep blues of the sea creates a stunning contrast that is both calming and peaceful.
Do you see those tiny houses? They can evoke a sense of coziness and warmth as it is like a layer under this view.
"""
```

```python
summary = summarizer(text, max_length=50, min_length=10)
```
